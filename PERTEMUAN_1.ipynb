**Tugas Praktikum D1**

```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

```python
# Membuat list untuk menyimpan url dari beberapa image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://cataas.com/cat"]
# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
  image = io.imread(url)                           #read image
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5) #resize image to half size
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)   #convert color to RGB
  final_frame = cv.hconcat((image, image_2))       #concatenate image
  cv2_imshow(final_frame)                          #show image
  print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ",tinggi," x ",lebar)
cv2_imshow(image_2)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat garis horizontal ditengah image
for y in range (lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

**Tugas Praktikum D2**

soal **2**

```python
image_vertikal = image_2.copy()
for x in range(tinggi):
    image_vertikal[x, int(lebar / 2)] = [255, 255, 255]

final_vertikal = cv.hconcat((image_2, image_vertikal))
cv2_imshow(final_vertikal)
print('\n')

image_diagonal = image_2.copy()
for i in range(min(tinggi, lebar)):
    y = int(i * (tinggi / min(tinggi, lebar)))
    x = int(i * (lebar / min(tinggi, lebar)))
    image_diagonal[y, x] = [255, 255, 255]


    image_diagonal[y, lebar - 1 - x] = [255, 255, 255]

final_diagonal = cv.hconcat((image_2, image_diagonal))
cv2_imshow(final_diagonal)
print('\n')
```

soal **3**

```python
image_horisontal_pendek = image_2.copy()

start_x = int(lebar * 0.25)
end_x = int(lebar * 0.75)
y_pos = int(tinggi / 2)

for y in range(start_x, end_x):
    image_horisontal_pendek[y_pos, y] = [255, 255, 255]

final_horisontal_pendek = cv.hconcat((image_2, image_horisontal_pendek))
cv2_imshow(final_horisontal_pendek)
print('\n')
```

soal **4**

```python
image_kotak = image_2.copy()

ukuran_kotak = 100
y_start = int(tinggi / 2) - int(ukuran_kotak / 2)
y_end = y_start + ukuran_kotak
x_start = int(lebar / 2) - int(ukuran_kotak / 2)
x_end = x_start + ukuran_kotak

image_kotak[y_start:y_end, x_start:x_end] = [255, 255, 255]

final_kotak = cv.hconcat((image_2, image_kotak))
cv2_imshow(final_kotak)
print('\n')
```
